```python
#group bar chart with matplotlib

import pandas as pd
import numpy
import matplotlib.pyplot as plt

dic1 = {'Product':['Apple','Banana','Beer','Juice','Apple','Banana','Beer','Juice','Apple','Banana','Beer','Juice'],
        'Year':[2000,2000,2000,2000,2001,2001,2001,2001,2002,2002,2002,2002],
        'Revenue':[1,3,4,5, 2,7,6,8, 3,6,2,9]}
df = pd.DataFrame(dic1)
df.head(10)

xsize=len(df['Year'].unique())#Number of unique years=3

x = numpy.arange(xsize)#[0 1 2]

bar_width = 0.15

fig,ax = plt.subplots()
ax.bar(x,'Revenue',width=bar_width,color='red',label="Apple",data=df[df['Product']=='Apple'])
ax.bar(x+bar_width,'Revenue',width=bar_width,color='yellow',label='Banana',data=df[df['Product']=='Banana'])
ax.bar(x+bar_width*2,'Revenue',width=bar_width,color="green",label='Beer',data=df[df['Product']=='Beer'])
ax.bar(x+bar_width*3,'Revenue',width=bar_width,color='blue',label='Juice',data=df[df['Product']=='Juice'])

ax.set_xticks(x+bar_width*2)
ax.set_xticklabels(['2000','2001','2002'])
ax.set_xlabel("Years")
ax.set_ylabel('Revenue')
ax.set_title("Grouped bar chart")
plt.legend()
plt.show()
```

```python
#Same above chart with seaborn
fig,ax = plt.subplots()
sns.barplot(x="Year",y="Revenue",hue="Product",data=df,estimator=sum,palette="muted",ax=ax)
plt.show()
```

```python
#Area charts

x = np.linspace(0,2,20)
fig,ax = plt.subplots()
ax.plot(x,x,label='linear',color='green')
ax.plot(x,x**3,label='cubic',color='red')

ax.fill_between(x,x,0,color='green',alpha=0.4)#0 represents y=0(x axis). Area to be filled between line and x axis
ax.fill_between(x,x**3,0,color='red',alpha=0.4)
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')
ax.set_title('Area chart')
ax.legend()
plt.show()
```

```python
x = np.linspace(0,2,20)
fig,ax = plt.subplots()
ax.plot(x,x,label='linear',color='green')
ax.plot(x,x**3,label='cubic',color='red')

ax.fill_between(x,x**3,x,color='yellow',alpha=0.4)#Area to be filled between lines
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')
ax.set_title('Area chart')
ax.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
#scatter chart
tips = sns.load_dataset('tips')
tips.head()

fig,ax = plt.subplots()
ax.scatter(x='total_bill',y='tip',color='green',marker='x',data=tips)
ax.set_xlabel('total bill')
ax.set_ylabel('tip')
ax.set_title('Scatter Chart')
plt.show()
```

```python
#scatter with third variable using colors to differentiate

#Creating a pandas series of colors using gender as mapping from tips data
colors = tips['sex'].map({'Male':'black','Female':'red'})
colors.head(10)

fig,ax = plt.subplots()
ax.scatter(x='total_bill',y='tip',color=colors,data=tips)
ax.set_xlabel('total bill')
ax.set_ylabel('tip')
ax.set_title('Scatter Chart')
plt.show()
```

```python
#Another example using colors with size and also varying the size of each point proportionally with size

tips['size'].unique()

colors = tips['size'].map({1:'blue',2:'red',3:'pink',4:'yellow',5:'green',6:'orange'})
sizes = tips['size']*20
fig,ax = plt.subplots()
ax.scatter(x='total_bill',y='tip',s=sizes,color=colors,data=tips)
ax.set_xlabel('total bill')
ax.set_ylabel('tip')
ax.set_title('Scatter Chart')
plt.show()
```

```python
#Varying the transparency of each point based on tip value

fig,ax = plt.subplots()
ax.scatter(x='total_bill',y='tip',c='tip',cmap='Reds',data=tips)
ax.set_xlabel('total bill')
ax.set_ylabel('tip')
ax.set_title('Scatter Chart')
plt.show()
```

```python
#Displaying a colorbar

fig,ax = plt.subplots()
im = ax.scatter(x='total_bill',y='tip',c='tip',cmap='plasma',data=tips)
fig.colorbar(im,ax=ax)

ax.set_xlabel('total bill')
ax.set_ylabel('tip')
ax.set_title('Scatter Chart')
plt.show()
```

```python
#Hexbin charts - Addresses the issue in scatter chart when data is huge and there are overlapping points

#Creating 10000 random points with normal distribution
n = 10000
df = pd.DataFrame({'x': np.random.randn(n), 'y': np.random.randn(n)})
df.head()
```

```python
#Using scatter plot to display a large amount of data. No clear info can be inferred.
fig,ax = plt.subplots(figsize=(7,7))
ax.scatter('x','y',data=df)
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(7,7))
im=ax.hexbin('x','y',gridsize=40,cmap='inferno',edgecolor='grey',data=df)
fig.colorbar(im,ax=ax)
plt.show()
```
