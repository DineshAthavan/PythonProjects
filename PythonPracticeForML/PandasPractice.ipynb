```python
s1.index#accessing index of the series
```

```python
s2 = pd.Series([5,10,15,20],index=['a','b','c','d'])#Using characters for index
s2
```

```python
s2.index
```

```python
s2['a']=7#changing the value at an index
s2
```

```python
s2['b':'d']#index slicing
```

```python
#Need of pandas series instead of lists -> Math operations
list1 = [2,4,6,8,10]
list2 = list1 + list1#instead of adding the individual elements in list, appending occurs
print(list2)
s3 = s1 + s1 #Adding two pandas series
s3#Mathematical operations can be performed with pandas series in an intuitive manner without any for loops
```

```python
#Options similar to matlab logical indexing is possible with pandas
filt = s2>10
print(filt)
print(s2[filt])
print(s2[s2>s2.mean()])#Using mean method with logical indexing
```

```python
#Creating a big series with same number
s3 = pd.Series(5,index=range(0,100))
s3
```

```python
#Creating a pandas series from a dictionary
profile = {'Id':123,'Name': 'abc','Age':35,'Role':'xyz'}
s4 = pd.Series(profile,name='Info')#key of dictionary is used as index, value of dice is used as values, name is used to mention meta data(optional)
s4
```

```python
#Creating a dataframe(2D tables) from dictionary
#Each column of the table is a pandas series
bio_dict = {'Name':['abc','def','ghi'], 'Age':[25,27,32], 'Id':[123,456,789]}
d1 = pd.DataFrame(bio_dict,index=['a','b','c'])
print(d1)
print(d1.index)
print(d1.columns)
```

```python
#accessing data from data frame
print(d1.Name)#accessing a column
print(d1['Name'])#accessing a column
print(d1[0:2])#accessing rows -> slicing is mandatory otherwise python thinks we are accessing columns and will lead to error
print(d1['Age'][0:1])#accessing subset of dataframe
print(d1.loc[['a'],['Name','Id']])#accessing subset using loc method. Characters based location selection.
print(d1.iloc[[1,2],[0,1]])#accessing subset using iloc method. Index based location selection.
```

```python
d2 = d1[d1['Age']>30]#Conditional filtering
d2
```
